```python
# Define chatbot intents and responses

intents = {
    "greeting": {
        "keywords": ["hi", "hello", "hey"],
        "response": "Hello! How can I help you today?"
    },
    "how_are_you": {
        "keywords": ["how are you", "how are you doing"],
        "response": "I'm just a chatbot, but I'm doing great! 😊"
    },
    "about_ai": {
        "keywords": ["what is ai", "artificial intelligence"],
        "response": "Artificial Intelligence is the simulation of human intelligence in machines."
    },
    "help": {
        "keywords": ["help", "what can you do"],
        "response": "I can answer basic questions and demonstrate a simple chatbot."
    },
    "goodbye": {
        "keywords": ["bye", "exit", "quit"],
        "response": "Goodbye! Have a nice day."
    }
}

print("Chatbot intents defined successfully")
```

```python
def get_response(user_input):
    user_input = user_input.lower()

    for intent in intents.values():
        for keyword in intent["keywords"]:
            if keyword in user_input:
                return intent["response"]

    return "Sorry, I didn't understand that. Can you rephrase?"
```

```python
print(get_response("hello"))
print(get_response("what is ai"))
print(get_response("bye"))
```

```python
conversation_log = []

print("Chatbot started. Type 'bye' to exit.\n")

while True:
    user_input = input("You: ")
    bot_response = get_response(user_input)

    conversation_log.append(f"You: {user_input}")
    conversation_log.append(f"Bot: {bot_response}")

    print(f"Bot: {bot_response}\n")

    if user_input.lower() in ["bye", "exit", "quit"]:
        break
```

```python
with open("chatbot_conversation.txt", "w") as file:
    for line in conversation_log:
        file.write(line + "\n")

print("Chatbot conversation saved successfully.")
```
